### Pre-requisites

```python
import sys
import os

def add_project_root_to_path(marker=".git"):
    """
    Adds the project root directory to sys.path based on the presence of a marker file or directory.

    Args:
        marker (str): The marker file or directory that indicates the project root. Default is '.git'.
    """
    current_dir = os.getcwd()

    def find_project_root(current_path, marker):
        while current_path != os.path.dirname(current_path):
            if marker in os.listdir(current_path):
                return current_path
            current_path = os.path.dirname(current_path)
        return None

    project_root = find_project_root(current_dir, marker)

    if project_root is None:
        raise RuntimeError("Project root not found")

    if project_root not in sys.path:
        sys.path.insert(0, project_root)

    print(f"Project root directory added to sys.path: {project_root}")


add_project_root_to_path()
```
